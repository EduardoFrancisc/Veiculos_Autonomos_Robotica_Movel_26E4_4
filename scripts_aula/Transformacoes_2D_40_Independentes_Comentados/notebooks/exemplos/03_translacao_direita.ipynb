# Exemplo 03 — Translacao positiva

O programa desloca a imagem 80 pixels para a direita e 30 para baixo por meio de uma matriz afim 2 × 3 aplicada com cv2.warpAffine. Ele imprime a matriz e compara visualmente a cena original com a transladada. A translação é útil para apresentar o conceito de deslocamento aparente da câmera e para estudar perturbações espaciais em pipelines de pré-processamento, lembrando que rótulos geométricos eventualmente associados à cena precisam acompanhar a mudança.
