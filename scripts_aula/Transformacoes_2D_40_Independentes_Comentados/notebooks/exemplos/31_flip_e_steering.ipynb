# Exemplo 31 — Flip horizontal e steering

O programa espelha horizontalmente a fotografia e mostra a regra didática de transformar um ângulo de direção θ em −θ quando o significado do rótulo e a geometria da tarefa admitem simetria. O código utiliza um valor de steering simulado, e não uma medição real do volante. O objetivo é ensinar que data augmentation supervisionado exige atualizar os rótulos de maneira coerente, evitando que a rede aprenda associações incorretas entre imagens e comandos.
