# Exemplo 18 — Interpolacao bilinear e bicubica

Este exercício compara as interpolações bilinear e bicúbica ao girar e ampliar uma fotografia, usando cv2.warpAffine com diferentes opções de reamostragem. Ambas estimam os valores dos pixels em posições transformadas, mas apresentam diferenças de nitidez e custo computacional. O exemplo é útil para discutir a qualidade das imagens fornecidas à CNN e os efeitos da reamostragem na preservação de detalhes.
