# Exemplo 33 — Aumento de dados por zoom

O programa aplica zooms de 0,9 e 1,1 vez à imagem frontal, demonstrando como o mesmo cenário pode aparecer com escalas ligeiramente distintas. As transformações são centradas e produzem resultados com bordas ou cortes diferentes. Esse tipo de augmentation pode ampliar a variabilidade visual do treino, mas deve ser utilizado com cuidado para não produzir cenas incompatíveis com a tarefa ou eliminar informações importantes da estrada.

```
from pathlib import Path
import zipfile, runpy, os

PACOTE = Path('/content/Transformacoes_2D_40_Independentes_Comentados')
if not PACOTE.exists():
    from google.colab import files
    arquivos = files.upload()  # selecione o arquivo ZIP completo desta aula
    nome = next((x for x in arquivos if x.lower().endswith('.zip')), None)
    if nome is None:
        raise ValueError('Envie o arquivo ZIP de exemplos')
    with zipfile.ZipFile(nome) as z:
        z.extractall('/content')
print('Pasta base:', PACOTE)
```
