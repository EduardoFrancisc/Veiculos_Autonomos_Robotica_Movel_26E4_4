# Exemplo 26 — Estatisticas RGB

O código converte uma imagem de BGR para RGB, coloca as intensidades no intervalo de zero a um e calcula a média e o desvio-padrão de cada canal. Esses indicadores descrevem parte da distribuição de brilho e cor das imagens. Eles são importantes porque a padronização de entradas de uma CNN depende das estatísticas escolhidas; em um dataset de treinamento real, devem ser calculadas somente a partir do conjunto de treino.
