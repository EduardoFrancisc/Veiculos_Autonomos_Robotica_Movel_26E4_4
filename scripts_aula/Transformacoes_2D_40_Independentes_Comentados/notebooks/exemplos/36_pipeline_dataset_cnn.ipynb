# Exemplo 36 — Pipeline integrado de imagens para CNN

Este programa resolve o exercicio integrado: carrega 20 imagens frontais com variações controladas, aplica letterbox para 224 × 224 e converte os pixels para RGB e valores entre zero e um. Antes da augmentação, separa treino e validação, calcula médias e desvios apenas no treino e aplica z-score. Gera versões com rotação de até 10 graus, zoom de até 10% e flip horizontal, registrando os parâmetros e invertendo o sinal apenas de rótulos demonstrativos simulados. Por fim, salva o dataset em NPZ, imagens processadas, metadados CSV, relatório e grids 4 × 5, permitindo apresentar todas as etapas exigidas pela atividade.

```
from pathlib import Path
import zipfile, runpy, os

PACOTE = Path('/content/Transformacoes_2D_40_Independentes_Comentados')
if not PACOTE.exists():
    from google.colab import files
    arquivos = files.upload()  # selecione o arquivo ZIP completo desta aula
    nome = next((x for x in arquivos if x.lower().endswith('.zip')), None)
    if nome is None:
        raise ValueError('Envie o arquivo ZIP de exemplos')
    with zipfile.ZipFile(nome) as z:
        z.extractall('/content')
print('Pasta base:', PACOTE)
```
