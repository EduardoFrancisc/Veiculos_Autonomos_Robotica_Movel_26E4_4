# Exemplo 32 — Aumento de dados por rotacao

O exemplo aplica duas rotações controladas, de menos 10 e mais 10 graus, e compara as imagens obtidas. Essas perturbações podem ajudar uma CNN a lidar com pequenas variações de orientação, câmera ou enquadramento. É importante esclarecer que uma rotação digital não equivale necessariamente a uma alteração física da trajetória, portanto rótulos reais de direção podem precisar de correções fundamentadas na geometria e na dinâmica do veículo.

```
from pathlib import Path
import zipfile, runpy, os

PACOTE = Path('/content/Transformacoes_2D_40_Independentes_Comentados')
if not PACOTE.exists():
    from google.colab import files
    arquivos = files.upload()  # selecione o arquivo ZIP completo desta aula
    nome = next((x for x in arquivos if x.lower().endswith('.zip')), None)
    if nome is None:
        raise ValueError('Envie o arquivo ZIP de exemplos')
    with zipfile.ZipFile(nome) as z:
        z.extractall('/content')
print('Pasta base:', PACOTE)
```
