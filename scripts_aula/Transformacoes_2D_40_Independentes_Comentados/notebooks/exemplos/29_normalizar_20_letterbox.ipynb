# Exemplo 29 — Normalizar um lote completo

O programa percorre as 20 fotografias do dataset, aplica letterbox 224 × 224 a cada uma e organiza o resultado em um único array NumPy. Além de salvar o grid 4 × 5, gera um arquivo compactado NPZ com as imagens normalizadas geometricamente. O exemplo demonstra como transformar arquivos separados em um conjunto de dados com dimensões consistentes, pronto para etapas posteriores de padronização e uso em CNNs.

```
from pathlib import Path
import zipfile, runpy, os

PACOTE = Path('/content/Transformacoes_2D_40_Independentes_Comentados')
if not PACOTE.exists():
    from google.colab import files
    arquivos = files.upload()  # selecione o arquivo ZIP completo desta aula
    nome = next((x for x in arquivos if x.lower().endswith('.zip')), None)
    if nome is None:
        raise ValueError('Envie o arquivo ZIP de exemplos')
    with zipfile.ZipFile(nome) as z:
        z.extractall('/content')
print('Pasta base:', PACOTE)
```

```
from pathlib import Path
import runpy
from IPython.display import Image, display

script = PACOTE / 'exemplos' / '29_normalizar_20_letterbox.py'
runpy.run_path(str(script), run_name='__main__')
resultados = PACOTE / 'saidas' / script.stem
for figura in sorted(resultados.glob('*.png'))[:2]:
    display(Image(filename=str(figura)))
```
