# Exemplo 27 — Padronizacao z-score

O exemplo aplica padronização z-score aos três canais RGB, subtraindo a média e dividindo pelo desvio-padrão de cada canal. O programa imprime as estatísticas antes e depois e produz uma representação visual apenas para inspeção, pois os valores padronizados deixam de corresponder diretamente às intensidades usuais de uma imagem. Essa transformação pode favorecer a estabilidade numérica do treinamento, desde que as estatísticas sejam estimadas no conjunto de treino e reutilizadas em validação e teste.
