# Exemplo 28 — Inspecao de conjunto com 20 imagens

O código carrega 20 imagens de câmera frontal com variações controladas de brilho, rotação e escala e monta um painel 4 × 5. Ele também consulta os metadados associados às imagens e confirma a quantidade e a resolução dos quadros. A importância desse exemplo está em verificar a diversidade e a integridade do conjunto de dados antes de iniciar normalização ou treinamento, identificando visualmente quais transformações já estão presentes nas amostras.

```
from pathlib import Path
import zipfile, runpy, os

PACOTE = Path('/content/Transformacoes_2D_40_Independentes_Comentados')
if not PACOTE.exists():
    from google.colab import files
    arquivos = files.upload()  # selecione o arquivo ZIP completo desta aula
    nome = next((x for x in arquivos if x.lower().endswith('.zip')), None)
    if nome is None:
        raise ValueError('Envie o arquivo ZIP de exemplos')
    with zipfile.ZipFile(nome) as z:
        z.extractall('/content')
print('Pasta base:', PACOTE)
```

```
from pathlib import Path
import runpy
from IPython.display import Image, display

script = PACOTE / 'exemplos' / '28_vinte_imagens_frontais.py'
runpy.run_path(str(script), run_name='__main__')
resultados = PACOTE / 'saidas' / script.stem
for figura in sorted(resultados.glob('*.png'))[:2]:
    display(Image(filename=str(figura)))
```
