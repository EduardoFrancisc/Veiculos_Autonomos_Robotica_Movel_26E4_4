# Exemplo 23 — Resize comum versus letterbox

O exemplo compara o redimensionamento direto de uma fotografia para 224 × 224 com o procedimento letterbox, que preserva a proporção original e completa as áreas restantes com padding. O primeiro método pode deformar carros, faixas e outros elementos da cena; o segundo mantém sua geometria relativa. Essa diferença é essencial para construir entradas padronizadas de CNNs sem introduzir distorções sistemáticas nos objetos.
