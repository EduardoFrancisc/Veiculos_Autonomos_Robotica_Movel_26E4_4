# Exemplo 30 — Mascarar a regiao de padding

O código cria uma máscara binária que distingue a área preenchida pela imagem original das faixas de padding adicionadas pelo letterbox. Ele calcula quantos pixels pertencem ao conteúdo válido e exibe a máscara ao lado da imagem normalizada. Esse procedimento ajuda a discutir como regiões artificiais podem afetar estatísticas, funções de perda e a interpretação dos dados quando uma CNN trabalha com fotografias de diferentes proporções.
