# Transformacoes 2D — 40 exemplos independentes
Execute uma atividade por vez. Todas as funcoes estao no arquivo Python escolhido; apenas as imagens sao compartilhadas.

```
from pathlib import Path
import zipfile, runpy, os

PACOTE = Path('/content/Transformacoes_2D_40_Independentes_Comentados')
if not PACOTE.exists():
    from google.colab import files
    arquivos = files.upload()  # selecione o arquivo ZIP completo desta aula
    nome = next((x for x in arquivos if x.lower().endswith('.zip')), None)
    if nome is None:
        raise ValueError('Envie o arquivo ZIP de exemplos')
    with zipfile.ZipFile(nome) as z:
        z.extractall('/content')
print('Pasta base:', PACOTE)
```

```
# Escolha 1 a 40
EXEMPLO = 3
codigos = sorted((PACOTE / 'exemplos').glob('[0-9][0-9]_*.py'))
assert len(codigos) == 40, 'Nao foram encontrados os 40 codigos'
caminho = codigos[EXEMPLO - 1]
print('Executando', caminho.name)
runpy.run_path(str(caminho), run_name='__main__')

from IPython.display import display, Image
pasta = PACOTE / 'saidas' / caminho.stem
for arq in sorted(pasta.glob('*.png'))[:2]:
    print('Resultado:', arq.name)
    display(Image(filename=str(arq)))
```
