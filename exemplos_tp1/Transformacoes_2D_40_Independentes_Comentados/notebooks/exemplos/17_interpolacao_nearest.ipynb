# Exemplo 17 — Interpolacao por vizinho mais proximo

O programa aplica a mesma transformação geométrica com interpolação por vizinho mais próximo e interpolação linear, exibindo as diferenças. O método nearest preserva valores discretos, sendo adequado para máscaras de segmentação que codificam classes por números inteiros. A comparação evidencia que o método de interpolação deve ser escolhido conforme o tipo de dado, pois suavizar uma máscara categórica pode criar valores que não representam classes válidas.
