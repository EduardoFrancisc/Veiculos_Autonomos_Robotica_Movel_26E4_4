# Exemplo 02 — Plano de coordenadas da imagem

O exemplo marca a origem das coordenadas, o centro e o canto inferior direito da fotografia, mostrando que no OpenCV o eixo x cresce para a direita e o eixo y cresce para baixo. O código calcula esses pontos a partir das dimensões da imagem e desenha indicadores sobre os pixels correspondentes. Compreender o sistema de coordenadas evita erros ao escolher o centro de rotação, calcular deslocamentos e transformar anotações de objetos em cenas de direção autônoma.
