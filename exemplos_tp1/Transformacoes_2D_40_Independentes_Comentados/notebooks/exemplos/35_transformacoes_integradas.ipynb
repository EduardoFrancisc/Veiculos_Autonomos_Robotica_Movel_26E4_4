# Exemplo 35 — Demonstracao integrada de transformacoes

Este é o programa integrador do exercicio integrado: sobre uma imagem frontal, ele executa translação, rotação em torno do centro, escala, matriz afim personalizada e a composição entre rotação e translação. Todas as matrizes são impressas no terminal, enquanto os resultados são salvos individualmente e reunidos em um painel comparativo. O exercício consolida as quatro transformações fundamentais e demonstra que a multiplicação de matrizes homogêneas permite organizar etapas de pré-processamento de forma matematicamente consistente.
