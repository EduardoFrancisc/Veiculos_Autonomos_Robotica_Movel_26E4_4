# Exemplo 16 — Transformacao das coordenadas de anotacoes

O código seleciona um ponto representativo da imagem e calcula sua nova posição por multiplicação com uma matriz afim que desloca a cena. Depois, desenha o ponto na fotografia original e na transformada, tornando visível a coerência entre imagem e coordenadas. Essa prática é indispensável ao transformar caixas delimitadoras, pontos de interesse e outras anotações utilizadas no treinamento supervisionado.
