# Exemplo 10 — Reflexao horizontal

Este exercício produz um espelhamento horizontal de duas maneiras: com cv2.flip e com uma matriz afim de reflexão aplicada por cv2.warpAffine. O código compara os resultados para demonstrar que ambos podem gerar a mesma inversão geométrica. O espelhamento é um recurso comum de data augmentation, mas em direção autônoma exige inverter também o sinal de um rótulo de steering quando a convenção e o cenário permitem essa simetria.
