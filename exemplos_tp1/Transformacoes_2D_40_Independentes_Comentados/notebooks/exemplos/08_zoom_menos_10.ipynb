# Exemplo 08 — Reducao digital de 10%

O programa reduz o conteúdo da imagem em 10%, utilizando um fator de escala igual a 0,9 em torno do centro. Como o tamanho final do canvas é mantido, surgem regiões de preenchimento nas bordas, tratadas pelo OpenCV. O exemplo permite discutir variações no campo visual, efeitos de borda e cuidados necessários para que a rede não aprenda artefatos artificiais produzidos durante o augmentation.

```
from pathlib import Path
import zipfile, runpy, os

PACOTE = Path('/content/Transformacoes_2D_40_Independentes_Comentados')
if not PACOTE.exists():
    from google.colab import files
    arquivos = files.upload()  # selecione o arquivo ZIP completo desta aula
    nome = next((x for x in arquivos if x.lower().endswith('.zip')), None)
    if nome is None:
        raise ValueError('Envie o arquivo ZIP de exemplos')
    with zipfile.ZipFile(nome) as z:
        z.extractall('/content')
print('Pasta base:', PACOTE)
```

```
from pathlib import Path
import runpy
from IPython.display import Image, display

script = PACOTE / 'exemplos' / '08_zoom_menos_10.py'
runpy.run_path(str(script), run_name='__main__')
resultados = PACOTE / 'saidas' / script.stem
for figura in sorted(resultados.glob('*.png'))[:2]:
    display(Image(filename=str(figura)))
```
