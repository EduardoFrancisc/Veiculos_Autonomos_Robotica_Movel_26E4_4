# Exemplo 19 — Tratamento das bordas

O código mostra duas estratégias para preencher áreas que aparecem fora dos limites originais após uma transformação: uma cor constante e a reflexão dos pixels próximos às bordas. As duas imagens resultantes são comparadas, destacando que o modo de borda afeta o aspecto visual da cena. Esse cuidado importa no data augmentation, pois bordas artificiais recorrentes podem se tornar pistas indesejadas para a rede neural.
