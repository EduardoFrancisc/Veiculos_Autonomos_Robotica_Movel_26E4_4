# Exemplo 05 — Rotacao em torno do centro

O código gira a fotografia 10 graus em torno do centro por meio de cv2.getRotationMatrix2D e cv2.warpAffine, exibindo a matriz de rotação resultante. O centro é calculado a partir da largura e da altura, o que evita que a imagem gire ao redor do canto superior esquerdo. Essa operação é utilizada em alinhamento de câmeras e em data augmentation para simular pequenas variações de orientação durante o treinamento.
