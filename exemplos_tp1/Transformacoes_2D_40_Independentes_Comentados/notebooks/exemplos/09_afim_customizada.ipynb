# Exemplo 09 — Matriz afim definida manualmente

O código constrói manualmente uma matriz afim 2 × 3 com componentes de escala, deslocamento e cisalhamento, sem depender de uma função pronta para gerar a matriz. Em seguida, aplica cv2.warpAffine e compara as imagens. O objetivo é mostrar que diferentes transformações geométricas podem ser representadas por uma única matriz, algo fundamental para alinhamento, registro de câmeras e preparação de imagens para CNNs.
