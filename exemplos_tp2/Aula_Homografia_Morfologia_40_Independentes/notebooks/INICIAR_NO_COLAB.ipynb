# Homografia, ORB, RANSAC e morfologia — 40 exemplos independentes

Execute as celulas: carregue o ZIP, selecione o programa e execute. Cada script tem codigo completo e resultados na pasta `saidas`. A versao Colab nao precisa de interface grafica.

```
import os, sys, zipfile, pathlib, subprocess
from google.colab import files
if not pathlib.Path('/content/aula/LEIA_PRIMEIRO.md').exists():
    arquivos = files.upload()
    nome = next((x for x in arquivos if x.lower().endswith('.zip')), None)
    if not nome: raise ValueError('Envie o ZIP de 40 exemplos.')
    with zipfile.ZipFile(nome) as z:
        z.extractall('/content')
subprocess.check_call([sys.executable, '-m', 'pip', '-q', 'install', 'opencv-python-headless', 'numpy', 'matplotlib'])
PASTA = pathlib.Path('/content/Aula_Homografia_Morfologia_40_Independentes')
assert PASTA.is_dir(), 'Pasta nao encontrada no ZIP'
print('Material:', PASTA)
```

```
# Selecione um numero de 1 a 40
NUMERO = 40
from IPython.display import display, Image
import glob, runpy
arquivos=sorted((PASTA / 'exemplos').glob(f'{NUMERO:02d}_*.py'))
assert len(arquivos)==1, 'Numero de exemplo invalido'
print('Executando:', arquivos[0].name)
runpy.run_path(str(arquivos[0]), run_name='__main__')
imgs=sorted((PASTA/'saidas'/arquivos[0].stem).glob('*.png'))
for img in imgs[:6]:
    print(img.name)
    display(Image(filename=str(img), width=950))
```
