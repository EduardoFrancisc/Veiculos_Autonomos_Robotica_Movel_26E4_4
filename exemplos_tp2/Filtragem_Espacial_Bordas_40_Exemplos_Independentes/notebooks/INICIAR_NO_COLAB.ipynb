```
from pathlib import Path
import zipfile, os
try:
    from google.colab import files
    print('Envie o arquivo ZIP completo que acompanha a aula.')
    enviados = files.upload()
    nome = next((x for x in enviados if x.lower().endswith('.zip')), None)
    if not nome: raise ValueError('É necessário enviar o ZIP dos 40 exercícios.')
    alvo = Path('/content/filtragem_espacial_aula')
    alvo.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(nome) as z:
        for item in z.infolist():
            p=(alvo/item.filename).resolve()
            if not p.is_relative_to(alvo.resolve()):
                raise ValueError('ZIP com caminho inválido')
        z.extractall(alvo)
    PASTA=alvo/'Filtragem_Espacial_Bordas_40_Exemplos_Independentes'
except ImportError:
    PASTA=Path.cwd()
    if not (PASTA/'exemplos').exists():
        raise RuntimeError('Execute o notebook na pasta extraída do ZIP.')
print('Pasta:',PASTA)
```

```
# Selecione o número do exercício, de 1 a 40.
EXEMPLO = 1
import runpy
from IPython.display import display, Image
scripts=sorted((PASTA/'exemplos').glob('*.py'))
if not (1<=EXEMPLO<=len(scripts)):
    raise ValueError('EXEMPLO deve estar entre 1 e 40')
arquivo=scripts[EXEMPLO-1]
print('Executando:',arquivo.name)
runpy.run_path(str(arquivo),run_name='__main__')
for figura in sorted((PASTA/'saidas'/arquivo.stem).glob('*.png'))[:5]:
    display(Image(filename=str(figura)))
```
