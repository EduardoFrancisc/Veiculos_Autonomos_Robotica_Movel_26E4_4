```python
from google.colab import files
from pathlib import Path
import zipfile
up=files.upload()
z=next((k for k in up if k.lower().endswith('.zip')),None)
if z is None: raise ValueError('Envie o ZIP desta aula')
base=Path('/content/meta_aula');base.mkdir(exist_ok=True)
with zipfile.ZipFile(z) as archive: archive.extractall(base)
roots=list(base.glob('**/exemplos'))
if not roots: raise RuntimeError('Pasta exemplos nao encontrada no ZIP')
root=roots[0].parent
print('Raiz:',root)
```
