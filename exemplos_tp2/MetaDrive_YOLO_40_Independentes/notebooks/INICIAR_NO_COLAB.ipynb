```
from google.colab import files
import zipfile
from pathlib import Path
uploaded=files.upload()
zip_name=next((x for x in uploaded if x.lower().endswith('.zip')),None)
if not zip_name: raise ValueError('Envie o ZIP')
base=Path('/content/metadrive_aula');base.mkdir(exist_ok=True)
with zipfile.ZipFile(zip_name) as z:z.extractall(base)
roots=list(base.glob('**/exemplos'));root=roots[0].parent
print(root)
```

```
import subprocess,sys
NUMERO=1
lista=sorted((root/'exemplos').glob(f'{NUMERO:02d}_*.py'))
assert len(lista)==1, 'Exemplo nao encontrado'
subprocess.run([sys.executable,str(lista[0])],cwd=root,check=True)
```
