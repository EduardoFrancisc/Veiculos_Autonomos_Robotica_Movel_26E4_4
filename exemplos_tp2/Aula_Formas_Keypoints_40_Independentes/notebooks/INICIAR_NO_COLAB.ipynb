# Visão computacional veicular: círculos, linhas, cantos e fluxo óptico
Abra este notebook no Google Colab. Execute a primeira célula, envie o ZIP da aula e depois escolha um número. Cada programa é independente e pode ser executado separadamente.

```
from google.colab import files
from pathlib import Path
import zipfile, subprocess, sys
up=files.upload()
z=next((Path(k) for k in up if k.lower().endswith('.zip')),None)
assert z is not None, 'Envie o ZIP do material.'
P=Path('/content/aula_formas_keypoints')
P.mkdir(exist_ok=True)
with zipfile.ZipFile(z) as a:
    for m in a.namelist():
        target=(P/m).resolve()
        assert target.is_relative_to(P.resolve()), 'Caminho inseguro no ZIP'
    a.extractall(P)
BASE=next(x for x in P.rglob('exemplos') if x.is_dir()).parent
print('Pasta:',BASE)
subprocess.run([sys.executable,'-m','pip','-q','install','opencv-python-headless','numpy','matplotlib'],check=True)
```

```
NUMERO = 40  # altere de 1 a 40
programas=sorted((BASE/'exemplos').glob('*.py'))
assert len(programas)==40
script=programas[NUMERO-1]
print('Executando:',script.name)
subprocess.run([sys.executable,str(script)],check=True)
```
